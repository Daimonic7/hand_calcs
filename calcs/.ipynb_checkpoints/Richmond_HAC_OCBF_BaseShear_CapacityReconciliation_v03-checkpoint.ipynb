## 1  Seismic Design Parameters (ASCE 7-22 / 2025 CBC)

Mapped and design spectral values from the Concept BOD (Site Class D, SDC D).
`SDS` governs the short-period plateau; `Ω0` and `Ie` follow Table 15.4-1 for a
Steel OCBF designed as a nonbuilding structure similar to a building.

```python
%%render params
S_S     = 1.960          # g, mapped $MCE_R$ (Ch. 22)
S_1     = 0.680          # g, mapped $MCE_R$
S_DS    = 1.310          # g, design short-period (governs)
S_D1    = 1.130          # g, design 1-second
R       = 3.250          # Steel OCBF, Table 15.4-1
Omega_0 = 2.000          # overstrength factor
I_e     = 1.000          # Risk Category II
T_a     = 0.182          # s, approx. period (Eq. 12.8-7)
W       = 54*kip         # module seismic weight (frame + MEP)
```

## 2  Period Region Check

The spectral corner period `Ts = SD1 / SDS` locates the module on the response
spectrum.

```python
%%render
T_s = S_D1 / S_DS          # spectral corner period (s)
```

Since `Ta = 0.182 s  ≪  Ts = 0.863 s`, the module sits on the **short-period
(constant-acceleration) plateau** — it attracts near-peak spectral acceleration.
This is the physical root of the acceleration problem the PBD study addresses.

## 3  Seismic Base Shear (ELF, ASCE 7-22 §12.8)

`Cs` from Eq. 12.8-2, bracketed by the upper limit (Eq. 12.8-3) and the two lower
limits (Eqs. 12.8-5, 12.8-6). Eq. 12.8-6 applies because `S1 = 0.68 g ≥ 0.6 g`.

```python
%%render
C_s     = S_DS / (R / I_e)              # Eq. 12.8-2  (governs)
C_smax  = S_D1 / (T_a * (R / I_e))      # Eq. 12.8-3  upper limit
C_smin1 = 0.044 * S_DS * I_e            # Eq. 12.8-5  lower limit
C_smin2 = 0.5 * S_1 / (R / I_e)         # Eq. 12.8-6  (S1 >= 0.6g)
V       = C_s * W                       # Eq. 12.8-1  base shear
```

`Cs = 0.403` by Eq. 12.8-2 (`0.105 < 0.403 < 1.910`, so neither limit
governs). **Seismic base shear `V = 21.8 kip` per 26-ft module** — this is the
permit-basis demand (`R = 3.25`) and does not change; everything below is the
beyond-code reconciliation.

## 4  Vertical Seismic Load Effect

```python
%%render
E_v = 0.2 * S_DS          # Eq. 12.4-4a  vertical seismic (x D)
```

## 5  Overstrength (Capacity-Design) Demand

Force-controlled demand for the brace connections and anchorage.

```python
%%render
V_o  = Omega_0 * V              # Eq. 12.4-7  force-controlled demand
V_mt = 1.5 * Omega_0 * V        # if multi-tiered OCBF (AISC 341 F1)
```

If the stacked-X frame is classified **multi-tiered** per AISC 341 §F1,
columns, struts, and their connections use `1.5 Ω0 = 3.0` (`Vmt = 65.3 kip`).
Resolve single- vs multi-tiered before finalizing the capacity-protection
checks — it moves the demand by 50%.

## 6  Demand Path — Brace Force & Base Reactions

Per-brace equilibrium at the base: the QE components `VQE` (vertical) and `HQE`
(horizontal) resolve into the brace axial `PQE`; the amplified reactions are the
anchorage / fuse sizing forces.

```python
%%render
V_QE  = 24*kip                       # per-brace vertical (uplift) QE
H_QE  = 9*kip                        # per-brace horizontal QE
theta = 69.4                         # deg, brace angle = atan($V_{QE} / H_{QE}$)
P_QE  = V_QE / sin(theta * pi/180)   # brace axial QE
P_u   = Omega_0 * P_QE               # amplified brace axial
H_u   = Omega_0 * H_QE               # amplified base shear (shear lug)
T_u   = Omega_0 * V_QE               # amplified base uplift (anchorage)
```

**Cross-check.** The amplified hand reactions (`Hu = 18 kip`, `Tu = 48 kip` per brace) match the S0.04 short-direction values (18 / 48), and the hand brace axial `Pu ≈ 51 kip` brackets RISA's lateral-brace `57 kip` — so the hand distribution is sane. **David's RISA base-end member forces (§9) govern the connection design**; the long-direction chevron (`C/T = 102.5 / 93.0 kip`) is the heaviest and sizes the fuse.

## 7  Brace Capacity (frame overstrength vs. demand)

Representative brace `HSS7×4×3/8` (confirm `Ag`, `r` from the member schedule /
AISC Manual). Expected tension yield per AISC 341; compression per AISC 360 §E3.

```python
%%render params
R_y  = 1.400             # Ry, expected/nominal yield ratio (A500 Gr.C HSS)
F_y  = 50*ksi            # nominal yield
A_g  = 6.900*inch**2     # gross area
K    = 1.000             # effective length factor
L_b  = 110.6*inch        # brace unbraced length
r_b  = 1.550*inch        # governing radius of gyration
E    = 29000*ksi         # modulus
```

```python
%%render
T_yield  = R_y * F_y * A_g                     # expected tension yield (AISC 341)
lambda_c = K * L_b / r_b                        # slenderness KL/r
F_e      = pi**2 * E / lambda_c**2              # Eq. E3-4  elastic buckling stress
F_cr     = 0.658**(F_y / F_e) * F_y             # Eq. E3-2  inelastic buckling
P_n      = F_cr * A_g                           # nominal compression capacity
```

Both brace capacities (`Tyield ≈ 483 kip`, `Pn ≈ 238 kip`) dwarf the
`≈ 26 kip` demand. The frame's true lateral overstrength is enormous, so a bare
(Solution 1) frame stays elastic to a very high base shear — which is exactly why
a base fuse pays off: it places a controlled yield far below the frame's brute
capacity and caps what reaches the servers.

## 8  Whole-Structure Overturning Check (per S. Aher, 2026-07-30)

§§1–7 give the ELF base shear **per 26-ft module** (permit basis, matches S0.04).
For the **fuse concept**, S. Aher directs an overturning check on the **whole
structure**: take the full HAC seismic weight from David's RISA dead load
(**≈ 230 kip** — the full 81′-6¼″ HAC, all main + side modules, not one module;
the bulk sits in the **interior five 9′-8″ bays** = 48′-4″ central, S3.01 / S3.03), apply the ELF shear
`V = Cs·W` as an equivalent lateral force at `0.7 h`, and compare the overturning
to the restoring available from self-weight. Geometry: transverse width and
height from **S3.04 / detail 1**; longitudinal bay layout from **S3.03**.
(Freebird's per-module takeoff was 54 kip; the RISA whole-HAC weight is the right
basis here.)

```python
%%render params
W_str = 230*kip          # whole-HAC seismic weight, RISA dead (81 ft-6.25 in; bulk in interior 5 bays)
h_str = 19.48*ft         # top of steel ~19.48 ft (19 ft 5-3/4 in), S3.04/1
k_h   = 0.700            # equivalent-force height factor (Sandesh: 0.7h; use 1.0 for full height)
B_ot  = 19.48*ft         # transverse base width, edge-to-edge (S3.04/1) - confirm
```

```python
%%render
V_str  = C_s * W_str            # whole-structure ELF base shear (Cs from Sec. 3)
h_eff  = k_h * h_str            # equivalent-lateral-force height
M_ot   = V_str * h_eff          # global overturning moment
M_rest = W_str * B_ot / 2       # restoring moment from self-weight (acts at center)
SR_ot  = M_rest / M_ot          # global stability ratio (> 1 => no net rigid-body uplift)
```

With `SR ≈ 1.8 > 1`, the HAC is **globally stable in rigid-body overturning** —
self-weight restoring is ~1.8× the overturning demand (still ~1.3× if the force is
applied at full height, `k_h = 1.0`). So the fuse tension is **not** set by global tipping; it is governed by **local
braced-frame uplift**, taken from David's RISA base-end forces in §9.
`V_str = Cs·W ≈ 93 kip` is the whole-HAC base shear (same either direction;
S. Aher's ~100–120 kip ballpark). Because the bulk of the 230 kip sits in the
**interior five bays**, the central **longitudinal** chevron frame carries most of
that inertia — consistent with the longitudinal chevron governing the base-end
forces (RISA `C/T = 102 / 93 kip`, §9).

## 9  Member End Forces at Base Connections (RISA)

Governing base-end member forces from David's RISA model, for the two base
details on S9.00 — **Detail 1: short (lateral) brace plate** and **Detail 5:
long (longitudinal) chevron brace base connection**. Sign convention axial
`+C / −T`; local **x = axial**, **y = lateral**, **z = longitudinal**. All shears
and `Myy` / torsion are negligible; only a small `Mzz` (about the longitudinal
axis) is carried. Representative members: lateral brace `HSS7×4×⅜` (S3.04),
longitudinal chevron `HSS6×3×⅜` (S3.03).

```python
%%render params
# axial: +C / -T  |  local x = axial, y = lateral, z = longitudinal
P_lat_C =  57.189*kip    # lateral (transverse) brace - axial compression  (Detail 1)
P_lat_T = -38.187*kip    # lateral (transverse) brace - axial tension      (Detail 1)
P_lon_C =  102.496*kip   # longitudinal chevron brace - axial compression  (Detail 5, governs)
P_lon_T = -93.037*kip    # longitudinal chevron brace - axial tension      (Detail 5, governs)
P_col_C =  25.886*kip    # braced-bay vertical column - axial compression
P_col_T = -15.222*kip    # braced-bay vertical column - axial tension
M_zz    = 4*kip*ft       # z-z connection moment (= 4 k-ft); all other V, M ~ 0
```

These supersede the hand distribution in §6 for connection design. The
**longitudinal chevron governs**: `C = 102.5 kip`, `T = 93.0 kip`. The lateral
brace carries `C/T = 57.2 / 38.2 kip` and the braced-bay column `25.9 / 15.2 kip`.

## 10  Base Connection & Fuse Rod — Forces & Sizing

The fuse is a **single ductile threaded rod** inline with the brace, tying the HSS
to a post-installed anchor base plate (Detail 1 lateral, Detail 5 longitudinal) and
acting as a pure **axial element** sized to the base-end tension. Base plates are
`PL 1¼″×18″×12″` with `¾″⌀ Gr. 105` anchor rods over a reinforced pier (S9.00/1, /5).
The fuse yields in tension; the base plate, anchors, gusset, and pier are
capacity-protected. The `z-z` moment adds a small tension couple across the anchor
rows; all other shears/moments are ≈ 0.

```python
%%render params
T_u   = 93.037*kip           # governing base-end tension (|P_lon_T|, long chevron - Detail 5)
T_lat = 38.187*kip           # lateral-brace tension (|P_lat_T| - Detail 1)
C_u   = 102.496*kip          # governing base-end compression (P_lon_C)
F_yr  = 105*ksi              # fuse-rod yield (F1554 Gr. 105, per S9.00)
phi_t = 0.900                # tension-yield resistance factor
A_pl  = 18*inch * 12*inch    # base-plate bearing area (PL 1-1/4 x 18 x 12, S9.00/1,5)
d_arm = 9*inch               # anchor-row lever arm for M_zz
```

```python
%%render
T_M   = M_zz / d_arm             # extra rod tension from the z-z moment couple
T_rod = T_u + T_M                # total governing fuse-rod tension
A_req = T_rod / (phi_t * F_yr)   # required reduced-shank area (tension yield)
d_req = sqrt(4 * A_req / pi)     # required reduced-shank diameter
f_br  = C_u / A_pl               # base-plate bearing stress (compression)
```

Governing fuse tension `T_rod ≈ 98 kip` (chevron `T = 93` + `Mzz` share `≈ 5`)
→ `d_req ≈ 1.15 in` → select a **1¼″⌀ reduced shank** (`φFy·A ≈ 116 kip ≥ 98`). The
lateral brace needs `T = 38 kip` → **¾″⌀ shank** (`φFy·A ≈ 42 kip`); the column
`T = 15 kip` → **½″⌀**. Base-plate bearing `f_br ≈ 0.47 ksi` ≪ concrete capacity —
not a governing check (vertical component governs in reality; still small).

**Capacity protection & scope.** The base plate, `¾″⌀ Gr. 105` anchors, gusset,
shear lug, and reinforced pier (S9.00/1, /5) are checked against the fuse
*overstrength* (`Ry·Rt` on `A·Fy`) so yielding stays in the rod. `Mzz = 4 k-ft` is
minor; all other shears/moments ≈ 0. Full base-plate / ACI 318-19 Ch. 17 anchorage
and pier design are the Week-2 50% package.

> **Confirm.** These are RISA base-end forces (governing / capacity combo). If they
> already include `Ω0`, the ductile fuse is sized to yield at the un-amplified
> demand (`÷Ω0 ≈ 49 kip → ⅞″⌀`) and only the protected hardware carries the full
> values — confirm the governing load combination with David. Gr. 105 matches the
> S9.00 anchors but is low-ductility; a milder turned-down shank (A36 / A572-50)
> may be preferable for fuse action.

## 11  Results Summary

```python
import pandas as pd
def kv(x, u="kip"):
    return f"{sig(float(x))} {u}".strip()
ct  = lambda c, t: f"{sig(float(c))} / {sig(abs(float(t)))} kip"
mft = lambda x: f"{sig(mag(x)/12.0)} kip\u00b7ft"   # moment in kip-ft (mag() = kip-inch)

rows = [
    ("Base shear - per module  V",                kv(V),          "R = 3.25 - permit basis"),
    ("Whole-structure base shear  Vstr",          kv(V_str),      "Wstr ~ 230 kip; interior 5 bays"),
    ("Global stability  SR = Mrest/Mot",          f"{sig(float(SR_ot))}\u00d7", "> 1 -> stable"),
    ("Lateral brace  C / T  (Detail 1)",          ct(P_lat_C, P_lat_T), "transverse"),
    ("Long chevron  C / T  (Detail 5, governs)",  ct(P_lon_C, P_lon_T), "longitudinal"),
    ("Braced-bay column  C / T",                  ct(P_col_C, P_col_T), ""),
    ("z-z moment  Mzz",                            mft(M_zz),      "all other V,M ~ 0"),
    ("Governing fuse tension  Trod",              kv(T_rod),      "T_chevron + Mzz share"),
    ("Fuse reduced shank  d_req",                 kv(d_req, "in"),"-> select 1-1/4 in dia"),
    ("Base-plate bearing  f_br",                  kv(f_br, "ksi"),"<< concrete capacity"),
]
df = pd.DataFrame(rows, columns=["Quantity", "Value", "Note"])
df
```

## 12  Reconciliation Targets for the 2D Perform-3D Model

The nonlinear model must reproduce this hand calc before any dynamic result is
trusted:

1. **Elastic period `T1 ≈ 0.18 s`** (modal check — if off, mass or stiffness is wrong).
2. **First yield occurs in the rod fuse**, at the base-end tension (lateral `≈ 38 kip`,
   longitudinal chevron `≈ 93 kip`); the telescoping shim interface transfers shear only.
3. **Pushover plateaus flat** (elastic–plastic axial fuse); a flag shape means
   unintended re-centering.
4. **Force ceiling `≤` fuse overstrength** (`Ry·Rt·A·Fy`) — base plate, anchors,
   gusset, shear lug, and pier stay elastic (capacity-design hierarchy).
5. **Global overturning stable** (`SR ≈ 1.8`); the model must not show rigid-body tipping.
6. **Reactions reconcile with RISA** at the base ends (`C/T` per §9) and with
   ACI 318-19 Ch. 17 anchorage in the 50% package.
