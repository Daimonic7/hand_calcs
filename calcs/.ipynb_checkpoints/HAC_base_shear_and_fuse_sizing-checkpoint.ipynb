# OAI Richmond Robotic Lab — HAC Base Shear & Threaded-Rod Fuse Sizing
**Preliminary hand calculations**  ·  *GPLA, Inc. — Structural Engineers and Builders*  ·  Phase 2

**Purpose.** Independent hand calc to (1) establish the HAC seismic **base shear** and (2) size the
rocking-base **threaded-rod fuse**, following the method S. Aher outlined on 2026-07-30:
`V = Cs·W` applied as a resultant at `0.7·h` → overturning; self-weight provides restoring;
the net overturning sets the fuse force. This is a **check on David's RISA model** — *not* a
permit calculation. Current permit design remains code-based (ELF, OCBF `R = 3.25`); the
rocking + fuse scheme is a **future-phase PBD options study**.

**References.** ASCE 7-22 (Ch. 12 & 15); AISC 360-22 / 341-22 (OCBF §F1); ACI 318-19 Ch. 17;
GPLA *HAC Seismic Criteria* (M. Fakhari, 2026-07-16); drawings S0.04, S3.03, S3.04, S9.00;
call w/ S. Aher, 2026-07-30.

**Sign convention.** Tension (uplift) positive for fuse demand. Items to verify are flagged **(CONFIRM)**.


```python
import math
try:
    import numpy as np
    import matplotlib.pyplot as plt
    HAVE_PLOT = True
except Exception:
    HAVE_PLOT = False

def rule(t=""):
    print("-" * 72)
    if t:
        print(t)
        print("-" * 72)

def show(name, val, unit="", w=32):
    print(f"{name:<{w}} = {val:>10.4g} {unit}")
```

## 1  Seismic design parameters  (ASCE 7-22 Ch. 15, nonbuilding *similar to* a building)

Steel **OCBF**: `R = 3.25`, `Ω0 = 2.0`, `Cd = 3.25`, RC II → `Ie = 1.0`, Site Class D / SDC D.
The seismic response coefficient `Cs` uses the Eq. 12.8-2 value bounded by the upper limit
(Eq. 12.8-3/4) and the Chapter-15 lower limits (Eq. 15.4-1/2). It should land at **≈ 0.402**.


```python
# --- ASCE 7-22 Ch.15 seismic inputs (GPLA criteria, M. Fakhari 2026-07-16) ---
Ss, S1        = 1.96, 0.68          # mapped MCE_R (g)
Fa, Fv        = 1.00, 2.50          # site coefficients (Site Class D)
R, O0, Cd, Ie = 3.25, 2.0, 3.25, 1.0   # Steel OCBF, nonbldg similar to bldg
TL            = 12.0                # s, long-period transition
Ct, xexp, hn  = 0.020, 0.75, 19.0  # ft, approximate-period parameters

SMS = Fa * Ss;  SM1 = Fv * S1
SDS = (2/3) * SMS;  SD1 = (2/3) * SM1
T   = Ct * hn ** xexp

# Seismic response coefficient Cs (ASCE 7-22 12.8.1 + 15.4.1 limits)
Cs_eq  = SDS / (R / Ie)                                        # Eq 12.8-2
Cs_max = SD1 / (T * (R / Ie)) if T <= TL else SD1 * TL / (T**2 * (R / Ie))  # Eq 12.8-3/4
Cs_m1  = max(0.044 * SDS * Ie, 0.03)                           # Eq 15.4-1 (>= 0.03)
Cs_mS1 = 0.8 * S1 / (R / Ie) if S1 >= 0.6 else 0.0             # Eq 15.4-2
Cs     = max(min(Cs_eq, Cs_max), Cs_m1, Cs_mS1)
Ev     = 0.2 * SDS                                             # vertical seismic effect

rule("SEISMIC PARAMETERS")
show("SMS", SMS, "g");   show("SM1", SM1, "g")
show("SDS", SDS, "g");   show("SD1", SD1, "g")
show("Approx. period T", T, "s")
show("Cs  Eq.12.8-2", Cs_eq);       show("Cs  upper limit", Cs_max)
show("Cs  min 0.044*SDS*Ie", Cs_m1); show("Cs  min S1>=0.6", Cs_mS1)
show("Cs  GOVERNING", Cs);           show("Ev = 0.2*SDS", Ev)
assert abs(Cs - 0.402) < 0.005, "Cs should be ~0.402"
```

## 2  Weight, geometry, load-combination bases

- `W = 230 kips` — total structure dead load from the RISA model (whole HAC: both main modules
  + side modules), **not** the 54-kip single Freebird module.
- Transverse bent (S3.04 Det. 1): telescoped fuses at **both** the columns (**8 ft** apart) and
  the diagonal-brace feet (**14.5 ft** apart). The outer brace feet are the compression-bearing
  **rocking pivot**, so the self-weight restoring lever is `14.5/2 = 7.25 ft`.
- Longitudinal (S3.03): a single chevron-braced bay (2nd interior bay), `B = 9'-8"`.

**Uplift dead-load factors (ASCE 7-22 §2.3.6).** Vertical seismic reduces the restoring dead load:
`kD_code = 0.9 − 0.2·SDS` (code strength) and `kD_mech = 1.0 − 0.2·SDS` (expected rocking
mechanism). Redundancy `ρ` multiplies the seismic overturning (check 1.0 vs 1.3).


```python
# --- Structure weight & geometry (RISA-confirmed / drawing dims) ---
W     = 230.0        # kips   total structure dead load (RISA)
h     = 19.42        # ft     top framing level (19'-5"); criteria hn = 19.0
h_eff = 0.70 * h     # ft     simplified ELF resultant height (per S. Aher)

# Transverse bent
d_col   = 8.0        # ft   column-to-column
d_brace = 14.5       # ft   brace-foot-to-brace-foot (outer bearing = rocking pivot)

# Longitudinal
B_long        = 9.667  # ft   braced bay width (9'-8")
frac_V_braced = 1.00   # -    fraction of longitudinal V into the braced bay (1.0 conservative)
P_D_col       = 14.0   # kips tributary DEAD at a braced-bay column   (CONFIRM from RISA)

# --- Load-combination dead-load factors for uplift ---
kD_code = 0.9 - 0.2 * SDS    # 0.9D - Ev   (code strength uplift combo)
kD_mech = 1.0 - 0.2 * SDS    # 1.0D - Ev   (expected rocking-mechanism check)
rho     = 1.0                # redundancy factor on Q_E (check 1.3)

rule("WEIGHT / GEOMETRY / COMBOS")
show("W (total dead)", W, "kips"); show("h", h, "ft"); show("0.7 h", h_eff, "ft")
show("column spread", d_col, "ft"); show("brace-foot spread", d_brace, "ft")
show("long. braced bay B", B_long, "ft"); show("P_D per braced-bay col", P_D_col, "kips")
show("kD code 0.9-0.2SDS", kD_code); show("kD mech 1.0-0.2SDS", kD_mech); show("rho", rho)
```

## 3  Base shear (ELF)

`V = Cs·W` (same magnitude both principal directions). The **elastic (unreduced)** base shear
`Ve = V·(R/Ie) ≈ SDS·W` is the demand the fuse ductility must absorb — carry it for context.


```python
V  = Cs * W               # design (reduced) base shear
Ve = V * (R / Ie)         # elastic (unreduced) ~ SDS*W

rule("BASE SHEAR")
show("V = Cs*W  (design)",   V,  "kips")
show("Ve = V*R/Ie (elastic)", Ve, "kips")
show("check SDS*W",          SDS * W, "kips")
```

## 4  Overturning → net uplift → fuse demand

Method (S. Aher): apply `V` at `0.7·h` to get the overturning moment `M_OT = ρ·V·(0.7h)`;
the self-weight restores about the compression pivot; the **net** overturning is picked up by the
tension-side fuses.

**Equilibrium about the compression pivot:**  `M_OT = M_restoring + Σ(T_fuse · d)`


### 4a  Transverse — rocking about the outer (leeward) brace foot
Restoring lever = `d_brace/2 = 7.25 ft`. The net (if any) is shared by the windward brace-foot
fuses along the length; `n_tens_trans` = number of those fuses (one per bent — adjust to layout).


```python
n_tens_trans    = 8              # windward brace-foot fuses along the length  (CONFIRM count)
lever_restore_T = d_brace / 2.0
M_OT_T          = rho * V * h_eff

def trans_fuse(kD, label):
    M_r   = kD * W * lever_restore_T
    net   = M_OT_T - M_r
    T_tot = max(net, 0.0) / d_brace         # total windward tension (resisted at brace feet)
    T_ea  = T_tot / n_tens_trans
    print(f"  [{label:<22}] M_OT={M_OT_T:7.0f}  M_restore={M_r:7.0f}  net={net:7.0f} k-ft"
          f"  ->  T_total={T_tot:5.1f} k   T/fuse={T_ea:5.1f} k")
    return T_ea

rule("TRANSVERSE OVERTURNING (pivot at outer brace foot, restoring lever 7.25 ft)")
Tt_code = trans_fuse(kD_code, "code 0.9D-Ev, rho=1.0")
Tt_mech = trans_fuse(kD_mech, "mech 1.0D-Ev, rho=1.0")
_net13  = 1.3 * V * h_eff - kD_code * W * lever_restore_T
print(f"  [code, rho=1.3         ] net={_net13:7.0f} k-ft"
      f"  ->  T/fuse={max(_net13,0)/d_brace/n_tens_trans:5.1f} k")
T_trans = max(Tt_code, Tt_mech)
```

### 4b  Longitudinal — single chevron-braced bay
The braced bay resists (conservatively) all of `V`; the overturning resolves as a
tension/compression couple over the bay width `B`. Gross column tension `= M_OT/B`, reduced by
the dead pin-down on that column.


```python
V_long  = frac_V_braced * V
M_OT_L  = rho * V_long * h_eff
T_gross = M_OT_L / B_long                    # gross tension at windward braced-bay column

def long_fuse(kD, label):
    T_net = T_gross - kD * P_D_col
    print(f"  [{label:<22}] M_OT={M_OT_L:7.0f} k-ft  T_gross={T_gross:6.1f} k"
          f"  pin-down={kD*P_D_col:5.1f} k  ->  T_fuse={T_net:6.1f} k")
    return max(T_net, 0.0)

rule("LONGITUDINAL OVERTURNING (single braced bay, B = 9'-8\")")
Tl_code = long_fuse(kD_code, "code 0.9D-Ev, rho=1.0")
Tl_mech = long_fuse(kD_mech, "mech 1.0D-Ev, rho=1.0")
_TL13   = 1.3 * V_long * h_eff / B_long - kD_code * P_D_col
print(f"  [code, rho=1.3         ]"
      f"                                        ->  T_fuse={_TL13:6.1f} k")
T_long  = max(Tl_code, Tl_mech)
```

```python
T_fuse = max(T_long, T_trans)
rule("GOVERNING FUSE TENSION")
show("Transverse   T/fuse",   T_trans, "kips")
show("Longitudinal T/fuse",   T_long,  "kips")
show(">> GOVERNING (design)", T_fuse,  "kips")
print("\nLongitudinal single-braced-bay uplift governs; the transverse fuses are")
print("minimum-/detailing-controlled (wide 14.5 ft base makes the transverse bent stable).")
```

## 5  Threaded-rod fuse sizing

The fuse is a **reduced-diameter smooth rod segment** spliced between couplers (per S. Aher).
It yields in tension on the reduced shank; couplers, larger rods, and everything downstream stay
elastic. Size the reduced-shank area so the **yield force ≈ demand** (`A = T/Fy`), then
capacity-protect all other elements to the fuse **expected** strength `Ry·Fy·A`.


```python
E     = 29000.0   # ksi
phi_t = 0.90      # LRFD tension-yield resistance factor

# (Fy, Fu, Ry) by grade -- confirm Ry per AISC 341 Table A3.1/A3.2 or product std
GRADES = {
    "A36 rod":      (36.0,  58.0, 1.5),
    "F1554 Gr.55":  (55.0,  75.0, 1.4),
    "F1554 Gr.105": (105.0, 125.0, 1.1),
}

def size_rod(T, Fy):
    A_nom  = T / Fy               # yield-force-based (fuse yields ~ at demand)
    A_lrfd = T / (phi_t * Fy)     # with phi
    return A_nom, math.sqrt(4*A_nom/math.pi), A_lrfd, math.sqrt(4*A_lrfd/math.pi)

rule(f"FUSE ROD SIZING   (T_demand = {T_fuse:.0f} kips)")
print(f"{'grade':<15}{'A_nom':>8}{'d_nom':>8}{'A_phi':>8}{'d_phi':>8}   expected Ry*Fy*A_nom")
for g, (Fy, Fu, Ry) in GRADES.items():
    A, d, Al, dl = size_rod(T_fuse, Fy)
    print(f"{g:<15}{A:8.2f}{d:8.2f}{Al:8.2f}{dl:8.2f}   {Ry*Fy*A:7.0f} kips")
print("\nA in in^2, d in in.  A_nom = T/Fy;  A_phi = T/(0.9*Fy).")
print("Yield on the reduced SMOOTH shank; keep couplers/threads/larger rods elastic.")
```

### 5b  Stretch (yield) length for ductility
The reduced-shank length between couplers is the yield length. Size it so the reduced shank
accommodates the fuse elongation demand `Δ` from rocking within a usable strain `ε` (with margin
below `ε_su`). `Δ` should come from the rocking kinematics / NLRHA — the value below is a
placeholder **(CONFIRM)**.


```python
delta_uplift = 0.75   # in    target fuse elongation at MCE  (CONFIRM from rocking/NLRHA)
eps_target   = 0.03   # in/in usable strain in the reduced shank (margin below eps_su)

grade_pick = "F1554 Gr.105"
Fy, Fu, Ry = GRADES[grade_pick]
d_fuse = 1.25         # in    reduced-shank diameter (rounded up from the table)
A_fuse = math.pi/4 * d_fuse**2
L_yield = delta_uplift / eps_target
dy      = Fy * L_yield / E
Fexp    = Ry * Fy * A_fuse

rule("FUSE DUCTILITY / STRETCH LENGTH")
print(f"  grade = {grade_pick},  d_fuse = {d_fuse} in,  A_fuse = {A_fuse:.2f} in^2")
show("fuse yield force  Fy*A", Fy * A_fuse, "kips")
show("fuse EXPECTED  Ry*Fy*A", Fexp, "kips")
show("target uplift  delta", delta_uplift, "in")
show("required yield length L", L_yield, "in")
show("elastic elong. at yield", dy, "in")
show("ductility  delta/dy", delta_uplift / dy, "")
```

### 5c  Capacity protection
Protect every other element to the fuse **overstrength** `F_protect = ω_sh · Ry · Fy · A`
(strain hardening `ω_sh ≈ 1.15`). Member properties below are placeholders **(CONFIRM from AISC
tables / the details)**.


```python
omega_sh = 1.15
F_prot   = omega_sh * Fexp
rule("CAPACITY-PROTECTION DEMAND (protect to fuse overstrength)")
show("F_protect = w_sh*Ry*Fy*A", F_prot, "kips")

# Larger (non-yielding) rod / coupler above the fuse
d_large = 1.75; A_large = math.pi/4 * d_large**2
phiRn_large = phi_t * Fy * A_large
print(f"  larger rod d={d_large}\" A={A_large:.2f}  phi*Fy*A = {phiRn_large:5.0f} k"
      f"  -> {'OK' if phiRn_large >= F_prot else 'NG'}  (>= {F_prot:.0f})")

# HSS brace tension yield (brace above the fuse) - AISC 360 D2
Ag_brace, Fy_brace = 6.18, 46.0        # HSS6x4x3/8 (CONFIRM Ag from AISC table)
phiPn_brace = 0.90 * Ag_brace * Fy_brace
print(f"  HSS brace phiPn (yield) = {phiPn_brace:5.0f} k"
      f"  -> {'OK' if phiPn_brace >= F_prot else 'NG'}")

print("  Concrete (ACI 318-19 Ch.17, 17.10 ductile option): design so")
print("    phi*Nsa (ductile fuse steel)  <  phi*Ncbg / Np / Nsb  (concrete)")
print("    develop breakout/pullout/side-face > F_protect with 9-10.5in embed,")
print("    pier #5 ties / #9 verticals, 12in mat.")
print("  Base plate flexure, knife plate & welds: design >= F_protect.")
print("  Shear (P*cos(theta)) taken by telescope + shims -> fuse stays in pure tension.")
```

### 5d  Required diameter vs. demand

```python
if HAVE_PLOT:
    Ts = np.linspace(20, 180, 60)
    plt.figure(figsize=(7, 4.2))
    for g, (Fy, Fu, Ry) in GRADES.items():
        plt.plot(Ts, np.sqrt(4 * (Ts / Fy) / np.pi), label=g)
    plt.axvline(T_fuse, ls="--", color="k", lw=1)
    plt.text(T_fuse + 2, 0.55, f"governing\n{T_fuse:.0f} k", fontsize=9)
    plt.xlabel("Fuse tension demand  T  (kips)")
    plt.ylabel("Required reduced-shank dia.  d  (in)")
    plt.title("Threaded-rod fuse: required diameter vs. demand  (A = T / Fy)")
    plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
else:
    print("matplotlib not available - skipping plot")
```
