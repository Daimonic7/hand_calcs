## 1  Seismic Design Parameters (ASCE 7-22 / 2025 CBC)

Design spectral values from the project **HAC Seismic Design Criteria** sheet (Site Class D, SDC D). $S_{DS}$ sets the short-period plateau; $\Omega_0$ and $I_e$ per Table 15.4-1 for a Steel OCBF designed as a nonbuilding structure similar to a building. $S_{DS} = 1.307$ g gives the locked $C_s = 0.402$.

```python
%%render params
S_S     = 1.960          # g, mapped $MCE_R$ (ASCE 7-22 Ch. 22)
S_1     = 0.680          # g, mapped $MCE_R$
S_DS    = 1.307          # g, design short-period spectral accel. (governs)
S_D1    = 1.133          # g, design 1-second spectral accel.
R       = 3.250          # Steel OCBF response modification factor (Table 15.4-1)
Omega_0 = 2.000          # overstrength factor for capacity-protected elements
I_e     = 1.000          # importance factor, Risk Category II
T_a     = 0.182          # s, approximate fundamental period (Eq. 12.8-7)
W       = 230*kip        # whole-HAC seismic weight (RISA dead)
h       = 19.48*ft       # top-of-steel height (S3.04/1)
```

## 2  Whole-Structure Base Shear (ELF, ASCE 7-22 §12.8)

Whole-HAC weight with the locked $C_s$. $V$ is the reduced design base shear; $V_e = V(R/I_e) = S_{DS} W$ is the elastic demand with no ductility. The gap between them is what the fuse ductility absorbs. Same magnitude in each principal direction.

```python
%%render
C_s = S_DS / (R / I_e)          # Eq. 12.8-2 seismic response coefficient (locked at 0.402)
V   = C_s * W                   # Eq. 12.8-1 design base shear, each principal direction
V_e = V * (R / I_e)             # elastic (unreduced) demand = $S_{DS} \cdot W$
```

With $T_a = 0.182$ s well below $T_s$, the HAC sits on the short-period plateau and draws near-peak spectral acceleration — the physical root of the server-acceleration problem the fuse addresses.

## 3  Vertical Seismic Load Effect

Reduces the gravity available for overturning restoring; governs the uplift combination in the stability check.

```python
%%render
E_v = 0.2 * S_DS          # Eq. 12.4-4a vertical seismic effect (fraction of dead load)
```

## 4  Overturning Demand (shared)

Apply the base shear $V$ as a single resultant at $0.7h$ (S. Aher simplification; use full $h$ for a conservative bound). $M_{ot}$ feeds the transverse stability check below.

```python
%%render params
k_h = 0.700              # equivalent-force height factor ($0.7h$ per S. Aher)
```

```python
%%render
h_eff = k_h * h                 # effective height of the lateral resultant
M_ot  = V * h_eff               # base overturning moment, each principal direction
```

# Lateral (Transverse) Direction

## 5  Geometry &amp; Load Distribution

The transverse system is a set of **base kickers** — no upper diaphragm or X-brace. Each of the **8 bents** (column lines 1-1 → 1-8) has **2 kickers**, one per column, splayed outboard from the 8 ft column line to feet 14.5 ft apart: **16 kickers total (8 per side), one ductile fuse per kicker**.

The kickers are the **only** transverse lateral system — there is no parallel moment frame in this direction, so they carry 100% of the transverse shear (no elastic-sharing relief, unlike the longitudinal line). Under a transverse push each bent develops one windward (tension) kicker and one leeward (compression) kicker sharing the bent shear; with no diaphragm each bent resists its own tributary inertia, so the bent over the widest tributary (adjacent to an end bay) governs.

```python
%%render params
n_bent    = 8              # transverse bents (column lines 1-1 through 1-8)
kick_bent = 2              # kickers per bent (one per column, splayed outboard)
b_foot    = 7.25*ft        # half of the kicker-foot spread (feet 14.5 ft apart)
run_k     = 3.25*ft        # kicker horizontal run, foot to column (S3.04/1)
rise_k    = 9.1*ft         # kicker vertical rise to work point (slope fit of S3.04/1, theta ~ 19.6 deg)
w_end     = 16.625*ft      # end-bay width (2 exterior bays)
w_int     = 9.667*ft       # interior-bay width (5 interior bays)
```

```python
%%render
n_kick  = n_bent * kick_bent            # total kickers = total lateral fuses
L_tot   = 2*w_end + 5*w_int             # overall HAC length (7 bays)
w_trib  = (w_end + w_int) / 2           # governing bent tributary width (bents 1-2, 1-7)
theta_k = atan(run_k / rise_k)          # kicker angle from vertical (rad)
L_k     = rise_k / cos(theta_k)         # kicker work-point length
```

Governing bent tributary width $=(w_{end}+w_{int})/2 = 13.15$ ft (bents 1-2, 1-7). Kicker angle $\theta_k \approx 19.6°$ from vertical — run 3.25 ft is dimensioned, rise ≈ 9.1 ft is a slope fit of both diagonals on S3.04/1 (left/right agree within 1°); confirm the work-point elevation against the model. Column bases taken as pinned, so the kickers carry the full transverse base shear at each bent.

```python
%%render
V_lat  = C_s * W                        # transverse base shear (whole HAC)
V_bent = V_lat * (w_trib / L_tot)       # shear at the governing bent (tributary, no diaphragm)
P_k    = V_bent / (2 * sin(theta_k))    # kicker axial = lateral fuse demand (windward tension)
```

## 6  Overturning Stability (global, transverse)

Whole-structure transverse overturning about a foot line, resisted by self-weight at the 7.25 ft lever, under the uplift-governing combination $0.9D - E_v$. Confirms the structure does not tip as a rigid body.

```python
%%render
M_rest  = W * b_foot                     # self-weight restoring at full dead (1.0D)
SR      = M_rest / M_ot                  # global stability ratio ($>1$ -> no rigid-body tipping)
W_up    = (0.9 - E_v) * W                # gravity in the uplift combination ($0.9D - E_v$)
M_netup = M_ot - W_up * b_foot           # net overturning beyond restoring
T_up    = M_netup / (2 * b_foot)         # total windward-line uplift (global; small)
```

$SR \approx 1.3 > 1$ — globally stable. The net rigid-body uplift $T_{up}$ over 8 windward feet is negligible per foot, so the **kicker axial $P_k \approx 22$ kip governs the lateral fuse**, not global overturning.

## 7  Lateral Fuse Sizing

Single ductile reduced-shank threaded rod inline with each kicker, sized to the kicker axial tension: $\phi_t F_y A \ge T_u$.

*Unit convention: structural lengths and moments are in ft / kip·ft; rod strength and member section properties are in ksi / in per steel-specification practice (rods are ordered by inch diameter; sections come from AISC tables).*

```python
# Lateral fuse demand carried forward as a plain kip magnitude for rod sizing.
T_u_lat = float(mag(P_k))    # kip
```

```python
%%render params
F_yr  = 105          # ksi, reduced-shank yield (F1554 Gr. 105)
phi_t = 0.900        # tension-yield resistance factor
```

```python
%%render
A_lat = T_u_lat / (phi_t * F_yr)     # reduced-shank area, lateral fuse [$\mathrm{in}^2$]
D_lat = sqrt(4 * A_lat / pi)         # reduced-shank diameter, lateral fuse [in]
```

$D_{lat} \approx 0.55$ in → **⅝″⌀ reduced shank** ($A = 0.307\ \mathrm{in}^2$, $\phi F_y A \approx 29\ \mathrm{kip} \ge 22\ \mathrm{kip}$): elastic at the code demand, yields below the MCE-elastic force — the intended fuse behavior. Capacity-protected base plate, anchors, gusset, and pier take $\Omega_0 P_k \approx 44$ kip (or fuse overstrength $R_y R_t \phi F_y A$), in the Week-2 50% package.

# Longitudinal Direction

## 8  System Behavior &amp; Force Path

The longitudinal line is a **stiffness hybrid**: one central X/chevron braced bay (S3.00, S3.03) works in parallel with welded-HSS moment frames in the other bays, tied to a common drift by the continuous columns and level beams (the beams also act as collectors, since there is no slab diaphragm).

For the **base fuse**, the governing force is the **base-diagonal axial from story shear** — a brace carries shear, so the fuse in it carries the braced bay's share of the longitudinal base shear resolved through the diagonal angle. Global overturning is resisted by the column **chords** over the full 81.6 ft footprint, not by the brace, so it is a separate anchorage check, not the brace-fuse demand. (v01's $M_{ot}/B_{bay}$ bracket conflated the two and dumped all overturning into one 9.7 ft bay — hence its inflated value.)

Because the braced bay and the moment frames act in parallel, they share the shear by relative stiffness: $\alpha_{br} = k_{br}/(k_{br}+k_{mf})$.

```python
%%render params
n_diag = 2               # braced-bay base diagonals (X to two base plates, S3.00 / S3.03)
run_b  = 4.0*ft          # base-diagonal horizontal run to work point (representative)
rise_b = 7.5*ft          # base-diagonal vertical rise to work point (representative)
```

```python
%%render
alpha_v = atan(run_b / rise_b)          # base-diagonal angle from vertical (~28 deg; confirm on S3.03/S3.00)
L_b     = rise_b / cos(alpha_v)         # base-diagonal work-point length
```

**Elastic shear sharing (approximate first-story stiffnesses, kip/in).** Braced bay = two diagonals in axial; moment frames = the other columns as portal columns. Section properties from AISC (diagonal HSS6×4×⅜, MF column HSS6×6×¼); first-story height 9′-3″. Base fixity is the main uncertainty — a pinned base would push $\alpha_{br}$ toward 0.95 (less relief).

```python
# Section properties & first-story geometry for the elastic-sharing estimate (ksi / in).
E_s   = 29000.0                     # ksi, steel modulus
A_d   = 6.18                        # in^2, HSS6x4x3/8 diagonal area
I_c   = 30.3                        # in^4, HSS6x6x1/4 moment-frame column
n_mfc = 14                          # moment-frame columns (16 total - 2 braced-bay)
h1_in = 9.25 * 12                   # in, first-story height (level @ 9'-3")
L_din = float(mag_ft(L_b)) * 12.0   # in, braced-bay diagonal length (from geometry above)
sinb  = float(sin(alpha_v))         # horizontal-component factor of the diagonal
```

```python
%%render
k_br     = n_diag * (E_s * A_d / L_din) * sinb**2     # braced-bay first-story stiffness [kip/in]
k_mf     = n_mfc * 12 * E_s * I_c / h1_in**3          # moment-frame columns, portal [kip/in]
alpha_br = k_br / (k_br + k_mf)                        # braced-bay share of longitudinal shear
```

The braced bay is ≈ 7× stiffer than the moment-frame sum, so the moment frames relieve only ≈ 12% ($\alpha_{br} \approx 0.88$). Elastic sharing helps, but it is a modest trim, not the main lever — the larger correction is the force-path reframe above.

```python
%%render
V_lon = C_s * W                           # longitudinal base shear (whole HAC)
V_br  = alpha_br * V_lon                   # shear to the braced bay after moment-frame sharing
P_lon = V_br / (n_diag * sin(alpha_v))     # base-diagonal axial = longitudinal fuse demand
```

## 9  Longitudinal Fuse Sizing

Same reduced-shank rod inline with the braced-bay base diagonal (S9.00/5), sized to the base-diagonal axial. Rod strength in ksi and diameter in inches, per the §7 convention.

```python
# Longitudinal fuse demand carried forward as a plain kip magnitude for rod sizing.
T_u_lon = float(mag(P_lon))    # kip
```

```python
%%render
A_lon = T_u_lon / (phi_t * F_yr)     # reduced-shank area, longitudinal fuse [$\mathrm{in}^2$]
D_lon = sqrt(4 * A_lon / pi)         # reduced-shank diameter, longitudinal fuse [in]
```

$D_{lon} \approx 1.08$ in → **1⅛″⌀ reduced shank** ($A = 0.994\ \mathrm{in}^2$, $\phi F_y A \approx 94\ \mathrm{kip} \ge 86\ \mathrm{kip}$); use **1¼″⌀** for margin pending confirmation. The force path (117 → 98 kip) plus elastic sharing (98 → 86 kip) brings the demand well below the old overturning bracket, and it reconciles with the RISA base-brace reaction (≈ 90 kip). Capacity-protected hardware takes $\Omega_0 P_{lon}$ (or fuse overstrength), in the 50% package.

## 11  Load Combinations

Both fuse demands are **lateral-shear-controlled brace axials** (seismic $E_h$). Gravity does not offset a brace axial, so no $0.9D - E_v$ term enters the fuse sizing — this resolves the earlier 1.0D-vs-$0.9D - E_v$ inconsistency, which came from the (now removed) overturning bracket. The $0.9D - E_v$ uplift combination is applied consistently in the **global overturning stability** check (§6, transverse). Longitudinal global overturning is non-critical: the 81.6 ft base length far exceeds the 19.5 ft height.

## 12  Reconciliation Targets for the 2D Perform-3D Model

1. Elastic period $T_1 \approx 0.18$ s (modal check).
2. First yield in the base fuse at the hand demand — lateral kicker $\approx 22$ kip, longitudinal base-diagonal $\approx 86$ kip (reconcile with the RISA base-brace reaction $\approx 90$ kip).
3. Pushover plateaus flat (elastic–plastic axial fuse); force ceiling at fuse overstrength $R_y R_t \phi F_y A$ so base plate, anchors, gusset, and pier stay elastic (capacity-design hierarchy).
4. Global transverse overturning stable ($SR \approx 1.3$ at full dead); net rigid-body uplift small.
5. Lateral shear spreads across the 16 kickers by tributary (no sharing); longitudinal shear concentrates in the one braced bay, relieved ~12% by the parallel moment frames — the residual lateral/longitudinal disparity is physical (16 fuses vs 2 base diagonals), not an artifact.
